here wo load a csv and turn it into a json for simpler usage. therefore we use the pandas library to read the csv file and turn it into a json file. the orient parameter creates a list of dictionaries (standard JSON Format), indent=2 makes the output human readable

```python
import pandas as pd
```

```python
df = pd.read_csv('assets/EHEC_oblig.csv')
df.to_json('output.json', orient='records', indent=2);
```
